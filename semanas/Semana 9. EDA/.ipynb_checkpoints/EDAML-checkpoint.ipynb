# Limpieza de datos
Filtramos nulos, vacíos y string con datos particulares

```python
from pyspark.sql.functions import col

# 1. OBLIGAMOS A SPARK A DISTINGUIR MAYÚSCULAS DE MINÚSCULAS
spark.conf.set("spark.sql.caseSensitive", "true")

# 2. CREAMOS df_renamed: Seleccionamos explícitamente las columnas originales
df_renamed = df_raw.select(
    col("`Titulo de Cargo`").alias("titulo_cargo"),
    col("`Empresa`").alias("empresa"),
    col("`Pais`").alias("pais"),
    col("`Fecha de Captura`").alias("fecha_captura"),
    col("`Descripcion`").alias("descripcion"),
    col("`Modalidad`").alias("modalidad"),
    col("`Tipo de Horario`").alias("tipo_horario"),
    col("`Fecha de Publicacion`").alias("fecha_publicacion")
)

# 3. CREAMOS df_cleaned: Filtramos las ofertas inválidas
df_cleaned = df_renamed.filter(
    (col("titulo_cargo").isNotNull()) & 
    (col("titulo_cargo") != "") & 
    (col("titulo_cargo") != "Sin titulo") &
    (col("titulo_cargo") != "No especificado") &
    (col("empresa").isNotNull()) &
    (col("empresa") != "")
)

# 4. Mostramos los resultados para verificar
df_cleaned.select("titulo_cargo", "empresa", "modalidad").show(10, truncate=False) 

print(f"Total de registros después de limpiar las ofertas inválidas: {df_cleaned.count()}")
```

# Extracción de datos de una columna

```python
from pyspark.sql.functions import col, regexp_extract

# 1. Extracción de Años de Experiencia requerida desde la Descripción
df_with_exp = df_cleaned.withColumn(
    "experiencia_raw", 
    regexp_extract(col("descripcion"), r'(?i)(\d+)\s*(?:año|ano)s?\s*de\s*experiencia', 1)
)

df_with_exp.select("titulo_cargo", "experiencia_raw").show(5, truncate=False)
```

# Normalización numérica
 Cambiamos la coma por punto y convertimos a Float para poder hacer clústeres

```python
from pyspark.sql.functions import col, udf
from pyspark.sql.types import IntegerType

# 2. UDF para normalizar la experiencia a números enteros
def normalizar_entero(valor):
    if not valor:
        return 0 # Si no especifica años, asumimos 0
    try:
        return int(valor)
    except:
        return 0

normalizar_udf = udf(normalizar_entero, IntegerType())

df_cleaned = df_with_exp.withColumn(
    "experiencia_anios", 
    normalizar_udf(col("experiencia_raw"))
).drop("experiencia_raw")
```

# Configuración de tipos de datos

```python
from pyspark.sql.functions import col, regexp_replace

# 3. Limpieza profunda de texto en la descripción
df_normalized = df_cleaned.withColumn(
    "descripcion_clean", 
    regexp_replace(col("descripcion"), r'[^a-zA-ZáéíóúÁÉÍÓÚñÑ0-9\s.,]', "")
).withColumn(
    "descripcion_clean", 
    regexp_replace(col("descripcion_clean"), r'\s+', " ") # Elimina espacios dobles
)

print("¡Limpieza de texto completada con éxito!")
```

```python
# 4. Casteo de Variables y Selección de Columnas (Mercado Laboral)
df_final = df_normalized.select(
    col("titulo_cargo"),
    col("empresa"),
    col("modalidad"),
    col("tipo_horario"),
    col("pais"),
    col("fecha_captura"),
    # Mantenemos la experiencia como Entero
    col("experiencia_anios").cast(IntegerType()).alias("experiencia_anios"),
    # Usamos la descripción ya limpia
    col("descripcion_clean").alias("descripcion")
)
```

```python
# 5. Previsualización de Datos Normalizados (Tipos Correctos):
print("--- PREVISUALIZACIÓN DE OFERTAS DE EMPLEO ---")
df_final.select("titulo_cargo", "empresa", "modalidad", "experiencia_anios").show(10, truncate=False)

# --- VERIFICACIÓN DE ESQUEMA ---
df_final.printSchema()
```

```python
from pyspark.sql.functions import col, when, concat, lit

df = df_final

# 1. Creación de la columna 'area_trabajo' buscando palabras clave en el título
df_enriquecido = df.withColumn(
    "area_trabajo",
    when(col("titulo_cargo").rlike("(?i)desarrollador|programador|software|datos|data|ingeniero|ti|it|sistema"), "Tecnología / IT")
    .when(col("titulo_cargo").rlike("(?i)venta|comercial|vendedor|ejecutivo|cuenta"), "Ventas / Comercial")
    .when(col("titulo_cargo").rlike("(?i)administracion|recursos humanos|rrhh|contable|finanza"), "Administración y Finanzas")
    .when(col("titulo_cargo").rlike("(?i)operario|logistica|bodega|conductor|chofer"), "Operaciones y Logística")
    .otherwise("Otras Áreas")
)

# 2. Crear un ID Único para diferenciar vacantes
# Esto combina Empresa + Área de Trabajo + Modalidad para evitar colisiones
df_final_ml = df_enriquecido.withColumn(
    "id_vacante_unico",
    concat(col("empresa"), lit("_"), col("area_trabajo"), lit("_"), col("modalidad"))
)

# 3. Mostrar los resultados para verificar la categorización
df_final_ml.select(
    "titulo_cargo", 
    "empresa", 
    "area_trabajo", 
    "experiencia_anios",
    "id_vacante_unico"
).show(10, truncate=False)
```

# Convertir Columnas a categóricas

```python
from pyspark.ml.feature import StringIndexer
```

1. Configurar el Indexador para la columna 'marca'

```python
from pyspark.ml.feature import StringIndexer

# handleInvalid="keep" es una buena práctica para no romper el pipeline si aparece una categoría nueva en el futuro
indexer_mod = StringIndexer(inputCol="modalidad", outputCol="modalidad_cat", handleInvalid="keep")
indexer_hor = StringIndexer(inputCol="tipo_horario", outputCol="tipo_horario_cat", handleInvalid="keep")
indexer_area = StringIndexer(inputCol="area_trabajo", outputCol="area_trabajo_cat", handleInvalid="keep")
```

2. Ajustar y transformar el DataFrame

```python
from pyspark.ml import Pipeline

# Agrupamos nuestros 3 indexadores en un solo proceso
pipeline = Pipeline(stages=[indexer_mod, indexer_hor, indexer_area])

# Esto crea las nuevas columnas '_cat' como un valor numérico (Double)
df_categorized = pipeline.fit(df_final_ml).transform(df_final_ml)
```

3. Limpieza final de columnas

```python
from pyspark.sql.functions import col

# Convertimos las columnas _cat a Integer para mayor claridad en el esquema
df_final_clustering = df_categorized.withColumn("modalidad_cat", col("modalidad_cat").cast("int")) \
                                    .withColumn("tipo_horario_cat", col("tipo_horario_cat").cast("int")) \
                                    .withColumn("area_trabajo_cat", col("area_trabajo_cat").cast("int"))
```

```python
# --- VERIFICACIÓN DE LOS DATOS ---
print(" Mapeo de Áreas de Trabajo a Categorías Numéricas:")
df_final_clustering.select("area_trabajo", "area_trabajo_cat").distinct().orderBy("area_trabajo_cat").show(truncate=False)

print(" Vista previa de los 10 registros listos para Machine Learning (Clustering):")
df_final_clustering.select(
    "id_vacante_unico", 
    "area_trabajo_cat", 
    "modalidad_cat", 
    "tipo_horario_cat", 
    "experiencia_anios"
).show(10, truncate=False)
```

# EDA Completo

## Paso 1: Análisis de Valores Faltantes

```python
from pyspark.sql.functions import count, when, isnan, col

df_eda = df_final_clustering

# Conteo de nulos por columna para verificar la calidad de las ofertas
print("--- CONTEO DE NULOS POR COLUMNA ---")
df_eda.select([count(when(isnan(c) | col(c).isNull(), c)).alias(c) for c in df_eda.columns]).show()
```

## Paso 2. Estadísticas Descriptivas (Numéricas)

```python
# Estadísticas para las variables cuantitativas y categóricas numéricas
print("--- ESTADÍSTICAS DESCRIPTIVAS (EXPERIENCIA Y CATEGORÍAS) ---")
df_eda.select(
    "experiencia_anios", 
    "area_trabajo_cat", 
    "modalidad_cat", 
    "tipo_horario_cat"
).describe().show()
```

## Estadísticas de Cardinalidad (Diversidad)

```python
from pyspark.sql.functions import avg, stddev, count, round

print("--- DEMANDA Y EXPERIENCIA PROMEDIO POR ÁREA DE TRABAJO ---")
df_eda.groupBy("area_trabajo") \
    .agg(
        count("*").alias("N_Vacantes"),
        round(avg("experiencia_anios"), 1).alias("Avg_Anios_Exp"),
        round(stddev("experiencia_anios"), 2).alias("Desviacion_Exp")
    ) \
    .orderBy(col("N_Vacantes").desc()) \
    .show(truncate=False)
```

## Identificación de Valores Atípicos (Outliers)

```python
# Detectar datos que podrían ensuciar el modelo.
# Ver vacantes que exigen niveles de experiencia irreales (sospechosos)
print("--- OUTLIERS: OFERTAS CON REQUISITOS DE EXPERIENCIA EXTREMOS ---")
df_eda.filter((col("experiencia_anios") > 20) | (col("experiencia_anios") < 0)).show(truncate=False)
```

## Paso 3. Conteo Global de Duplicados
Un duplicado exacto es aquel donde toda la fila se repite.

```python
total_registros = df_eda.count()
unicos_registros = df_eda.distinct().count()

print(f"Total de ofertas registradas: {total_registros}")
print(f"Ofertas únicas reales: {unicos_registros}")
print(f"Ofertas duplicadas eliminables: {total_registros - unicos_registros}")
```

## Guardar el progreso (Checkpoint)

```python
# Crear una vista temporal para usar SQL
df_eda.createOrReplaceTempView("mercado_laboral_clean")

# Consultas SQL: Promedio de experiencia exigida según la Modalidad (Presencial, Remoto...)
print("--- EXPERIENCIA PROMEDIO POR MODALIDAD DE TRABAJO ---")
spark.sql("""
    SELECT 
        modalidad, 
        ROUND(AVG(experiencia_anios), 1) as avg_experiencia 
    FROM mercado_laboral_clean 
    GROUP BY modalidad 
    ORDER BY avg_experiencia DESC
""").show()
```

```python
# Consulta avanzada para ver concentración de vacantes por empresa
print("--- TOP EMPRESAS CONTRATANDO POR ÁREA ---")
spark.sql("""
    SELECT 
        empresa, 
        area_trabajo,
        ROUND(AVG(experiencia_anios), 1) as exp_promedio,
        COUNT(*) as cantidad_vacantes
    FROM mercado_laboral_clean 
    GROUP BY empresa, area_trabajo 
    ORDER BY cantidad_vacantes DESC, empresa ASC
""").show(20, truncate=False)
```
