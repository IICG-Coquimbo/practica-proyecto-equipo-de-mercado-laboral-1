```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

url = "https://ineed.cl"

headers = {
    "User-Agent": "Mozilla/5.0"
}

response = requests.get(url, headers=headers)
soup = BeautifulSoup(response.text, "html.parser")

jobs = []

# 🔍 buscar textos generales (más robusto para esta web)
for item in soup.find_all(["div", "article", "section"]):
    texto = item.get_text(separator="\n").strip()

    if len(texto) < 50:
        continue

    titulo = texto.split("\n")[0]

    pais = "Chile" if "Chile" in texto else None

    modalidad = None
    if "Remoto" in texto:
        modalidad = "Remoto"
    elif "Híbrido" in texto:
        modalidad = "Híbrido"
    elif "Presencial" in texto:
        modalidad = "Presencial"

    horario = None
    if "Full" in texto:
        horario = "Full Time"
    elif "Part" in texto:
        horario = "Part Time"

    moneda = None
    if "$" in texto:
        moneda = "CLP/USD"

    jobs.append({
        "Título": titulo,
        "País": pais,
        "Modalidad": modalidad,
        "Fecha": None,
        "Moneda": moneda,
        "Categoría": None,
        "Horario": horario
    })

df = pd.DataFrame(jobs)

df = df.drop_duplicates()
df = df[df["Título"].notna()]

print(df.head())

df.to_csv("trabajos_ineed.csv", index=False, encoding="utf-8-sig")

print("✅ LISTO → trabajos_ineed.csv")
```
