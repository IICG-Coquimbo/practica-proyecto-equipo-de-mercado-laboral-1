```
1. ¿El problema es el algoritmo o los datos?

El problema son los datos, no el algoritmo. El modelo de regresión lineal falla 
porque se intentó predecir el precio del alimento basándose únicamente en la cantidad de estrellas 
y el número de opiniones. El precio de un producto veterinario no se asigna por cuántos 
comentarios tenga en la web, por lo que la línea de predicción se queda plana. El algoritmo hizo su trabajo 
matemático, pero la realidad del mercado "golpeó" al modelo por usar variables equivocadas.
```

```
2. ¿Cómo lo solucionarían en el mundo real sin hacer trampa? ¿Qué datos faltan en el dataset original que sí 
determinen el precio de un producto veterinario?

Para solucionarlo en el mundo real sin hacer trampa, se debe mejorar el dataset 
añadiendo variables que sí tengan un impacto directo en la estructura de costos del producto.

El costo de los ingredientes o materia prima.
El tamaño o peso neto del empaque.
El posicionamiento de la marca.
El origen del producto.
El margen de ganancia que aplica la tienda de mascotas.
```
