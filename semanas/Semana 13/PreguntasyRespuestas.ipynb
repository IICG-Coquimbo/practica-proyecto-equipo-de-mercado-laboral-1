¿El problema es el algoritmo o los datos?

El fallo se debe a los datos, ya que la regresión lineal funciona bien, pero se alimentó con información irrelevante; las estrellas de rating o la cantidad de comentarios no tienen una relación real ni lógica con el precio de un producto


¿Cómo lo solucionarían en el mundo real sin hacer trampa? qué datos faltan
en el dataset original que sí determinen el precio de un producto veterinario
(por ejemplo: costo de la materia prima, margen de ganancia de la tienda,
si es importado o nacional, peso neto del empaque, etc.). O en sus datos,
como fué el algoritmo?

La solución en el mundo real consiste en cambiar los datos actuales por variables comerciales que sí determinen el costo, tales como el peso neto del empaque, la calidad de los alimentos tanto como los mas economicos y los mas costosos, si el producto es importado o nacional, y el margen de ganancia de la tienda
