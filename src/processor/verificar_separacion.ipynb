# Validación de Separación Contenedor A / B
Verifica que raw_data y processed_data existen en Atlas con registros correctos.

```python
from pymongo import MongoClient
import os
from dotenv import load_dotenv

load_dotenv()

client = MongoClient(os.getenv("MONGO_URI"))
db = client[os.getenv("DB_NAME", "Proyecto_Bigdata")]

raw       = "Registros_Scraping"
processed = "processed_data"

n_raw       = db[raw].count_documents({})
n_processed = db[processed].count_documents({})

print("=" * 45)
print("VALIDACION SEPARACION CONTENEDOR A / B")
print("=" * 45)
print(f"raw_data       ({raw}): {n_raw} documentos")
print(f"processed_data ({processed}): {n_processed} documentos")
print("=" * 45)

assert n_raw > 0,       "ERROR: raw_data esta vacio"
assert n_processed > 0, "ERROR: processed_data esta vacio"
assert n_processed <= n_raw, "ERROR: processed_data tiene mas registros que raw_data"

print("OK: Separacion correcta entre Contenedor A y Contenedor B")
client.close()
```
