```python
import os
from pymongo import MongoClient
import pandas as pd

# 1. Leer las credenciales ocultas que inyectaste desde el .env
MONGO_URI = os.getenv("MONGO_URI")
# Se actualizan los valores por defecto a los nombres reales de tu Atlas
DB_NAME = os.getenv("DB_NAME", "Proyecto_Bigdata")
COLLECTION_RAW = os.getenv("COLLECTION_RAW", "Registros_Scraping")
COLLECTION_PROCESSED = os.getenv("COLLECTION_PROCESSED", "Registros_Limpios")

# 2. Establecer la conexión segura con MongoDB Atlas
client = MongoClient(MONGO_URI)
db = client[DB_NAME]

print("Conectado a Atlas exitosamente.")

# 3. Extraer los datos crudos (Scraper)
coleccion_raw = db[COLLECTION_RAW]
datos_crudos = list(coleccion_raw.find())

# Convertir a DataFrame de Pandas para que tu equipo haga la limpieza
df = pd.DataFrame(datos_crudos)

# =========================================================
# AQUÍ TUS COMPAÑEROS DEBEN ESCRIBIR SU CÓDIGO DE LIMPIEZA
# (Eliminar nulos, outliers, crear 3 columnas nuevas, etc.)
# =========================================================

# 4. Guardar los datos limpios en la nueva colección (Regla estricta)
coleccion_limpia = db[COLLECTION_PROCESSED]

# Se convierte el DataFrame limpio de vuelta a formato diccionario para Mongo
df_limpio = df.to_dict('records')
coleccion_limpia.insert_many(df_limpio)

# Se actualiza el print para que sea dinámico y muestre el nombre real de tu colección
print(f"Datos limpios guardados exitosamente en la colección: {COLLECTION_PROCESSED}")
```
